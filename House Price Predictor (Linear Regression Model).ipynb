```python
# Importing libraries needed for all the three algorithms.
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
# Importing libraries needed for Linear Regression model.
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
# Reading the dataset into a dataframe df.
df=pd.read_csv("smart_house_prices.csv")
# Preprocessing dataset. X (inputs) are all columns except the target and Y (output) is the target.
X=df.drop(columns=["Price_INR"])
Y=df["Price_INR"]
X=pd.get_dummies(X, drop_first=True)
# Splitting dataset into training data and testing data.
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
# Training the Linear Regression model.
model = LinearRegression()
model.fit(X_train, Y_train)
# Making predictions of the house prices for the test set.
Y_pred= model.predict(X_test)
# Evaluating the model.
mae=mean_absolute_error(Y_test, Y_pred)
mse=mean_squared_error(Y_test, Y_pred)
rmse=np.sqrt(mse)
r2= r2_score(Y_test, Y_pred)
print(f"Mean Absolute Error = {mae:,.2f}")
print(f"Mean Squared Error = {mse:,.2f}")
print(f"Root Mean Squared Error = {rmse:,.2f}")
print(f"R2 Score = {r2:,.2f}")
```

```python
# House price prediction for a new input using linear regression model.
new_house = {
    "Size_sqft": 1500,
    "Rooms": 3,
    "Bathrooms": 2,
    "Location": "Delhi",
    "Age_yrs": 5,
    "Furnishing": "Semi-Furnished",
    "House_Type": "Apartment",
    "Parking": 1
}
new_df = pd.DataFrame([new_house])
new_df = pd.get_dummies(new_df, drop_first=True)
new_df = new_df.reindex(columns=X.columns, fill_value=0)
predicted_price = model.predict(new_df)
print(f"Predicted Price: {predicted_price[0]:,.2f}")
```
